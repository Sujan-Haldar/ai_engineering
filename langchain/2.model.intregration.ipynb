### Streaming

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
reply = model.stream("Write a 2000 line paragraph in AI")
for chunk in reply:
    print(chunk.text,end='\n',flush=False)
```

### Batch
Batching collection of independent request to a model can significantly improve performance and reduce costs, as the processing can be done parallel.


```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
responses = model.batch(["What is java?","Do you know everything"])
for response in responses:
    print(response)
```
